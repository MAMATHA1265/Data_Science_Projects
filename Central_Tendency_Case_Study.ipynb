```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
#create  a histogram of "fixed acidity" feature
plt.figure(figsize=(10,5))
sns.histplot(data=df, x='fixed acidity',color='blue',edgecolor='linen',alpha=0.5,bins=5)
plt.title('Histogram of Fixed Acidity')
plt.xlabel('Fixed Acidity')
plt.ylabel('Frequency')
plt.show()
```

```python
#calculate the mean of 'fixed Acidity' feature
mean_fixed_acidity=df['fixed acidity'].mean()
print('Mean of Fixed Acidity:',mean_fixed_acidity)
```

```python
#calculate the median of 'fixed Acidity' feature
median_fixed_acidity=df['fixed acidity'].median()
print('Median of Fixed Acidity:',median_fixed_acidity)
```

```python
#create  a histogram of "fixed acidity" feature and also show mean and median
plt.figure(figsize=(10,5))
sns.histplot(data=df, x='fixed acidity',color='blue',edgecolor='linen',alpha=0.5,bins=5)
plt.title('Histogram of Fixed Acidity')
plt.xlabel('Fixed Acidity')
plt.ylabel('Frequency')
plt.axvline(mean_fixed_acidity,color='red',linestyle='dashed',linewidth=2,label='Mean')
plt.axvline(median_fixed_acidity,color='green',linestyle='dashed',linewidth=2,label='Median')
plt.legend()
plt.show()
```

```python
#create histogram of "volatile acidity" feature
plt.figure(figsize=(10,5))
sns.histplot(data=df, x='volatile acidity',color='green',edgecolor='linen',alpha=0.5,bins=5)
plt.title('Histogram of Volatile Acidity')
plt.xlabel('Volatile Acidity')
plt.ylabel('count')
plt.show()
```

```python
#plot distribution of "Volatile Acidity" feature
plt.figure(figsize=(10,5))
sns.kdeplot(data=df,x='volatile acidity',color='pink',fill=True)
plt.title('Distribution of Volatile Acidity')
plt.xlabel('Volatile Acidity')
plt.ylabel('Density')
plt.show()
```

```python
#create skewness of "Volatile Acidity" feature
data=df['volatile acidity']
skewness=data.skew()
print('Skewness of Volatile Acidity:',skewness)
```

```python
#create  a histogram of "Volatile Acidity" feature also show mean and median
plt.figure(figsize=(10,5))
sns.histplot(data=df, x='volatile acidity',color='green',edgecolor='linen',alpha=0.5,bins=5)
plt.title('Histogram of Volatile Acidity')
plt.xlabel('Volatile Acidity')
plt.ylabel('count')
plt.axvline(mean_fixed_acidity,color='red',linestyle='dashed',linewidth=2,label='Mean')
plt.axvline(median_fixed_acidity,color='green',linestyle='dashed',linewidth=2,label='Median')
plt.legend()
plt.show()
```

```python
# create a histogrsm od"Citric Acid" feature
plt.figure(figsize=(10,5))
sns.histplot(data=df, x='citric acid',color='orange',edgecolor='linen',alpha=0.5,bins=5)
plt.title('Histogram of Citric Acid')
plt.xlabel('Citric Acid')
plt.ylabel('count')
plt.show()
```

```python
from matplotlib import legend
##create  a histogram of "Citric acid" feature and also show mean and median
plt.figure(figsize=(10,5))
sns.histplot(data=df, x='citric acid',color='orange',edgecolor='linen',alpha=0.5,bins=5)
plt.title('Histogram of Citric Acid')
plt.xlabel('Citric Acid')
plt.ylabel('count')
plt.axvline(mean_fixed_acidity,color='red',linestyle='dashed',linewidth=2,label='Mean')
plt.axvline(median_fixed_acidity,color='green',linestyle='dashed',linewidth=2,label='Median')
plt.legend()
plt.show()
```

```python
#create a distribution "Citric Acid" feature
plt.figure(figsize=(10,5))
sns.kdeplot(data=df,x='citric acid',color='orange',fill=True)
plt.title('Distribution of Citric Acid')
plt.xlabel('Citric Acid')
plt.ylabel('Density')
plt.show()
```

```python
#create count plot of "Quality" feature
plt.figure(figsize=(10,5))
sns.countplot(data=df,x='quality',color='purple',edgecolor='linen',alpha=0.5)
plt.title('Count Plot of Quality')
plt.xlabel('Quality')
plt.ylabel
```

```python
#count the number of occurance of different category of "quality" feature
category_counts=df['quality'].value_counts()
print('Category Counts of Quality:')
print(category_counts)
```

```python
#calculate mode of "quality" feature
mode_quality=df['quality'].mode()
print('Mode of Quality:',mode_quality)
```

```python
#create new pandas dataframe called "rec_acid" that contains the different representative averages of different types of acid
data=df.groupby('quality')[['fixed acidity','volatile acidity','citric acid']].mean()

rec_acid = data.rename(columns={
    'fixed acidity': 'Fixed Acidity',
    'volatile acidity': 'Volatile Acidity',
    'citric acid': 'Citric Acid'
})
rec_acid.index.name = 'Quality'

print(rec_acid)
```
