```python
# create scatter plot of "age" and "salary" feature
plt.figure(figsize=(10, 6))
sns.scatterplot(x="age", y="salary", data=data)
plt.title("Scatter Plot of Age vs. Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
# create scatter plot of 'healthy_eating', and 'active_lifestyle', feature
plt.figure(figsize=(10, 6))
sns.scatterplot(x="healthy_eating", y="active_lifestyle", color='red', edgecolor='blue', data=data)
plt.title("Scatter Plot of Healthy Eating vs. Active Lifestyle")
plt.xlabel("Healthy Eating")
plt.ylabel("Active Lifestyle")
plt.show()
```

```python
#create scatter plot of 'healthy_eating', and 'salary', feature
plt.figure(figsize=(10, 6))
sns.scatterplot(x="healthy_eating", y="salary", color='green', edgecolor='blue', data=data)
plt.title("Scatter Plot of Healthy Eating vs. Salary")
plt.xlabel("Healthy Eating")
plt.ylabel("Salary")
plt.show()
```

```python
#create a count plot of "groups" feature
plt.figure(figsize=(10, 6))
sns.countplot(x="groups", data=data, edgecolor='linen', alpha=0.7)
plt.title("Count Plot of Groups")
plt.xlabel("Groups")
plt.ylabel("Count")
plt.show()
```

```python
# create a histogram of "salary" feature
plt.figure(figsize=(11, 6))
sns.histplot(x="salary", data=data, color='orange',edgecolor='linen',bins=10,alpha=0.7)
plt.title("Histogram of Salary")
plt.xlabel("Salary")
plt.ylabel("count")
plt.show()
```
